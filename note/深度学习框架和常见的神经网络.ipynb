### 深度学习

机器学习是实现人工智能的一种途径。
深度学习是机器学习的一个子集,也就是说深度学习是实现机器学习的一种方法。
深度学习是机器学习中一种基于对数据进行特征学习的算法。
深度学习是基于**人工神经网络**,深度是指网络中使用多层,每层都通过非线性变换处理数据,并逐渐提取出更复杂、更抽象的特征。

**核心原理**

**从感知机到神经网络**

感知机（Perceptron） 是最简单的神经网络模型，它接收多个输入信号，为每个输入分配权重，将所有加权信号累加并加上偏置，最后通过激活函数决定是否输出。

单层感知机的局限性在于无法解决非线性可分问题（如XOR异或问题），这催生了多层感知机（MLP）的发展。

(深度depth learning 允许多个神经元进行计算，单个神经元只能做线性决策，而多层网络可以逐层提取越来越抽象的特征——底层识别边缘，中层识别形状，高层识别完整物体。)

Machine Learning 机器学习<br>
Input ->Feature extraction(Men) **->** Classification -> output

Deep Learning 深度学习<br>
Input ->Feature extraction(Auto) **+** Classification -> output

传统机器学习算法依赖人工设计特征,并进行特征提取;而深度学习方法不需要人工,而是依赖算法自动提取特征。
深度学习通过模仿人脑的神经网络来处理和分析复杂的数据,从大量数据中自动提取复杂特征。<br>
*这也是深度学习被看做黑盒子,可解释性差的原因。*<br>
深度学习尤其擅长处理高维数据,如图像、语音和文本。<br>

些运算、处理的方法被封装在类中。

PyTorch张量与NumPy数组类似,但PyTorch的张量具有GPU加速的能力(通过CUDA),这使得深度学习模型能够高效地
在GPU上运行。

````python
import torch

# 创建一个 2D 张量
tensor = torch.tensor([[1, 2, 3], [4, 5, 6]], dtype=torch.float32)
print("原始张量:\n", tensor)


print("获取第一行:", tensor[0])  # 获取第一行
print("获取第一行第一列的元素:", tensor[0, 0])  # 获取特定元素
print("获取第二列的所有元素:", tensor[:, 1])  # 获取第二列所有元素


reshaped = tensor.view(3, 2)  # 改变张量形状为 3x2
print("改变形状后的张量:\n", reshaped)
flattened = tensor.flatten()  # 将张量展平成一维
print("展平后的张量:\n", flattened)

tensor_add = tensor + 10  # 张量加法
print("张量加 10:\n", tensor_add)
tensor_mul = tensor * 2  # 张量乘法
print("张量乘 2:\n", tensor_mul)
tensor_sum = tensor.sum()  # 计算所有元素的和
print("张量元素的和:", tensor_sum.item())

tensor2 = torch.tensor([[1, 1, 1], [1, 1, 1]], dtype=torch.float32)
print("另一个张量:\n", tensor2)
tensor_dot = torch.matmul(tensor, tensor2.T)  # 张量矩阵乘法
print("矩阵乘法结果:\n", tensor_dot)
````

````python
import torch
import numpy as np

# 1. NumPy 数组转换为 PyTorch 张量
numpy_array = np.array([[1, 2, 3], [4, 5, 6]])
print("NumPy 数组:\n", numpy_array)

# 使用 torch.from_numpy() 将 NumPy 数组转换为张量
tensor_from_numpy = torch.from_numpy(numpy_array)
print("转换后的 PyTorch 张量:\n", tensor_from_numpy)

# 修改 NumPy 数组
numpy_array[0, 0] = 100
print("修改后的 NumPy 数组:\n", numpy_array)
print("PyTorch 张量也会同步变化:\n", tensor_from_numpy)

# 2. PyTorch 张量转换为 NumPy 数组
print("\n2. PyTorch 张量转为 NumPy 数组")
tensor = torch.tensor([[7, 8, 9], [10, 11, 12]], dtype=torch.float32)
print("PyTorch 张量:\n", tensor)

# 使用 tensor.numpy() 将张量转换为 NumPy 数组
numpy_from_tensor = tensor.numpy()
print("转换后的 NumPy 数组:\n", numpy_from_tensor)

# 共享内存
tensor[0, 0] = 77
print("修改后的 PyTorch 张量:\n", tensor)
print("NumPy 数组也会同步变化:\n", numpy_from_tensor)

# 不共享内存的情况（需要复制数据）
tensor_independent = torch.tensor([[13, 14, 15], [16, 17, 18]], dtype=torch.float32)
numpy_independent = tensor_independent.clone().numpy()  # 使用 clone 复制数据
print("原始张量:\n", tensor_independent)
tensor_independent[0, 0] = 0  # 修改张量数据
print("修改后的张量:\n", tensor_independent)
print("NumPy 数组（不会同步变化）:\n", numpy_independent)
````

### 自动求导

````python
import torch

# 创建一个张量，并指定需要计算梯度
x = torch.tensor([11.0, 2.0, 3.0, 14.0], requires_grad=True) #requires_grad==True PyTorch 才会在计算图中跟踪它们的操作

# 对张量进行操作
y = 4 * x +  x ** 2  # dy = 4 + 2*x

#反向传播计算梯度

y.sum().backward()

print(x.grad_fn)
print(x.grad)
````

````python
#多变量自动求导
import torch

x1 = torch.tensor(1.0, requires_grad=True)
x2 = torch.tensor(2.0, requires_grad=True)
y = x1 ** 2 + x2 ** 3

y.backward()

print(x1.grad)
print(x2.grad)
````

### 神经网络
是一种模仿人脑处理信息方式的计算模型，它由许多相互连接的节点（神经元）组成，这些节点按层次排列。

神经网络的强大之处在于其能够自动从大量数据中学习复杂的模式和特征，无需人工设计特征提取器。

随着深度学习的发展，神经网络已经成为解决许多复杂问题的关键技术。

神经元（Neuron）
神经元是神经网络的基本单元，它接收输入信号，通过加权求和后与偏置（bias）相加，然后通过激活函数处理以产生输出。

神经元的权重和偏置是网络学习过程中需要调整的参数。

输入和输出:

输入（Input）：输入是网络的起始点，可以是特征数据，如图像的像素值或文本的词向量。<br>
输出（Output）：输出是网络的终点，表示模型的预测结果，如分类任务中的类别标签。<br>
神经元接收多个输入（例如x1, x2, ..., xn），如果输入的加权和大于激活阈值（activation potential），则产生二进制输出。<br>


#### 层（Layer）
输入层和输出层之间的层被称为隐藏层，层与层之间的连接密度和类型构成了网络的配置。

神经网络由多个层组成，包括：

输入层（Input Layer）：接收原始输入数据。
隐藏层（Hidden Layer）：对输入数据进行处理，可以有多个隐藏层。
输出层（Output Layer）：产生最终的输出结果。

##### 全连接网络（MLP）<br>
MLP适用于结构化数据（如表格数据）、小型图像分类（如MNIST）和简单回归问题。然而，由于参数量大、对局部模式不敏感，MLP在处理高维数据（如图像或序列）时表现有限，容易过拟合且训练深层网络时可能面临梯度消失问题。
##### RNN（循环神经网络，Recurrent Neural Network）
核心作用：专为处理序列数据设计。其核心思想是引入“记忆”机制，在处理当前时间步的数据时，会结合上一时间步的隐藏状态（历史信息），从而捕捉数据中的时序依赖关系。<br>
适用领域/场景：自然语言处理、语音识别、时间序列预测（如股票、天气）、音乐生成、机器人控制等。
##### LSTM（长短期记忆网络，Long Short-Term Memory）
核心作用：RNN的一种特殊变体。为了解决传统RNN在处理长序列时容易出现的“梯度消失”或“梯度爆炸”问题而设计。它通过引入“细胞状态”和三个门控机制（遗忘门、输入门、输出门），能够选择性地记忆长期重要信息并遗忘无用信息。<br>
适用领域/场景：机器翻译、文本生成、情感分析、语音识别、长时间跨度的时间序列预测等需要捕捉长距离依赖的序列任务。
##### GRU（门控循环单元，Gated Recurrent Unit）
核心作用：LSTM的简化版本。它将LSTM的遗忘门和输入门合并为“更新门”，并新增“重置门”，去除了独立的细胞状态。在保持与LSTM相近性能的同时，大幅减少了参数数量，降低了计算复杂度，训练速度更快。<br>
适用领域/场景：适用于资源受限（如移动端、嵌入式系统）、数据规模较小或序列长度适中的自然语言处理及时间序列预测任务。
##### GAN（生成对抗网络，Generative Adversarial Network）
核心作用：由“生成器”和“判别器”两个相互博弈的神经网络组成。生成器负责生成逼真的假数据，判别器负责区分真假数据，两者在对抗训练中共同提升，最终使生成器能够创造出以假乱真的新数据。<br>
适用领域/场景：图像生成与修复、人脸生成、图像风格迁移、数据增强、超分辨率重建等生成式任务。
##### GNN（图神经网络，Graph Neural Network）
核心作用：专门用于处理图结构数据（由节点和边组成）。它能够通过节点之间的连接关系传播和聚合信息，学习图中实体及其相互关系的表示。<br>
适用领域/场景：社交网络分析、推荐系统、分子结构预测（计算生物学）、交通流量预测、知识图谱推理等具有复杂关系网络的任务。
##### Transformer
核心作用：完全基于“自注意力机制（Self-Attention）”，摒弃了RNN的递归结构。它能够并行处理整个输入序列，并灵活捕捉序列中任意两个位置之间的长距离依赖关系，极大提升了训练效率和模型性能。<br>
适用领域/场景：自然语言处理（机器翻译、大语言模型、文本摘要）、计算机视觉（Vision Transformer, ViT）、多模态学习（如CLIP）等，是目前深度学习领域的主流架构。
##### Autoencoder（自编码器）
核心作用：一种无监督学习网络，包含“编码器”和“解码器”。编码器将输入数据压缩为低维的潜在表示（Latent Representation），解码器再尝试从该表示中重构出原始输入。其核心目的是学习数据的高效压缩表示或提取核心特征。<br>
适用领域/场景：数据降维、异常检测（如工业设备故障检测）、图像去噪、特征提取以及作为生成模型的基础（如变分自编码器 VAE）。

##### 卷积神经网络（CNN，Convolutional Neural Network）
核心作用：专门用于处理具有网格结构的数据（如图像）。它通过卷积层自动提取局部特征（如边缘、纹理），并通过池化层降低数据维度，具有平移不变性和参数共享的优势，能够高效地进行层级化特征学习。<br>
CNN 是计算机视觉任务（如图像分类、目标检测和分割）的核心技术。<br>
一个标准的CNN通常由 特征提取器（卷积层+池化层） 和 分类器（全连接层） 串联构成：<br>

● 卷积层：在局部区域用卷积核做加权和，提取边缘、纹理等局部特征；参数共享，相比全连接参数量更少。<br>
● 池化层：对特征图做下采样（如取窗口内最大值），降维、减少计算量，并对小位移有一定鲁棒性。<br>
● 全连接层：将前面得到的特征展平后做分类或回归，与前面章节中的全连接用法一致。<br>
典型流程为：卷积 → 激活 → 池化（可重复多段），最后接全连接输出。<br>

*[输入图像] → [卷积层+激活] → [池化层] → [卷积层+激活] → [池化层] → ... → [展平] → [全连接层] → [输出]*<br>
适用领域/场景：计算机视觉（图像分类、目标检测、图像分割、人脸识别）、自动驾驶视觉感知、医学影像分析，以及语音识别和自然语言处理中的文本分类等。

````python
import torch
import torch.nn as nn
from torchvision.datasets import CIFAR10
from torchvision.transforms import ToTensor
import torch.optim as optim
from torch.utils.data import DataLoader
import time
import matplotlib.pyplot as plt
from torchsummary import summary


# 加载数据集转换成张量数据集
def create_dataset():
    train_dataset = CIFAR10(root='./data', train=True, transform=ToTensor())
    valid_dataset = CIFAR10(root='./data', train=False, transform=ToTensor())# ToTensor(): 将图片数据转换成张量数据
    return train_dataset, valid_dataset

class ImgModel(nn.Module):
	def __init__(self):
		super().__init__()
		# 输入图像为 32x32。两次卷积（无填充）后：

		# 卷积 1：3 -> 6，kernel=3，stride=1，padding=0。输出尺寸 = 32 - 3 + 1 = 30。

		# 池化 1：kernel=2，stride=2。输出尺寸 = 30 / 2 = 15。

		# 卷积 2：6 -> 16，kernel=3，stride=1，padding=0。输出尺寸 = 15 - 3 + 1 = 13。

		# 池化 2：kernel=2，stride=2。输出尺寸 = 13 // 2 = 6

		# 接着展平为 16 * 6 * 6 = 576

		#输入3，输出6，卷积核大小3*3，步长1，填充0
		self.conv1 = nn.Conv2d(3, 6, 3, 1, 0)
	
		self.pool1 = nn.MaxPool2d(2, 2, 0)
		#卷积层
		self.conv2 = nn.Conv2d(6, 16, 3, 1, 0)
		
		self.pool2 = nn.MaxPool2d(2, 2, 0)
		# 池化层

		self.linear1 = nn.Linear(16 * 6 * 6, 120)
		# 隐藏层

		self.linear2 = nn.Linear(120, 84)
		# 输出层

		# 10分类
		self.out = nn.Linear(84, 10)

	def forward(self, x):
		# 卷积+激活+池化
		x = self.pool1(torch.relu(self.conv1(x)))
		
		# 2 卷积+激活+池化
		x = self.pool2(torch.relu(self.conv2(x)))

		
		x = x.reshape(shape=(x.shape[0], -1))

		x = torch.relu(self.linear1(x))

		# 第2层隐藏层
		x = torch.relu(self.linear2(x))

		# 输出层
		x = self.out(x)
		return x
	
BATCH_SIZE = 8               # 批量大小
EPOCHS = 1                   # 训练轮数
LEARNING_RATE = 0.01          # 学习率
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
def train(train_dataset):
	#加载数据
	train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
	#加载对象
	model = ImgModel().cuda()
	#损失函数
	criterion = nn.CrossEntropyLoss()
	#优化器
    #optimizer = optim.SGD(model.parameters(), lr=LEARNING_RATE, momentum=0.9)
	optimizer = optim.Adam(model.parameters(), lr = 1e-3)

	for epoch in range(EPOCHS):
		total_loss, total_samples, total_correct, start = 0.0, 0.0, 0.0, time.time()
		#训练模式
		model.train()
		for x, y in train_loader:
			x, y = x.to(DEVICE), y.to(DEVICE)
			#预测
			y_pred = model(x)
			#计算损失
			loss = criterion(y_pred, y)
			#清零
			optimizer.zero_grad()
			# 梯度计算
			loss.backward()
			# 参数更新
			optimizer.step()
			
			# 统计预测正确的样本个数
			total_correct += ((torch.argmax(y_pred, 1) == y).sum())#预测为true的数量
			#print(y_pred)
			#print(torch.argmax(y_pred, 1))
			
			# loss.item(): 当前批次平均损失值
			total_loss += loss.item() * len(y)# 统计当前批次的总损失值
			
			total_samples += len(y)# 统计当前批次的样本数
		end = time.time()
		print('轮数:%2s loss:%.5f acc:%.2f time:%.2fs' % (
			epoch + 1, total_loss / total_samples, total_correct / total_samples, end - start))
	# 保存训练模型
	torch.save(obj=model.state_dict(), f='./model/imagemodel.pth')

def test(test_dataset):
	# test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=True)
    # model = ImgModel().cuda()
    # model.load_state_dict(torch.load('./model/imagemodel.pth'))

if __name__ == '__main__':
	train_dataset, test_dataset = create_dataset()
	# print('train_dataset.data.shape->', train_dataset.data.shape)
	# 图像
	# plt.figure(figsize=(2, 2))
	# plt.imshow(train_dataset.data[1])
	# plt.title(train_dataset.targets[1])
	# plt.show()
	model = ImgModel().cuda() #需要指定在GPU
	#summary(model, input_size=(3, 32, 32))

# Conv2d-1 (168个)：(输入通道3 * 卷积核高3 * 卷积核宽3 + 1个偏置) * 输出通道6 = (3*3*3 + 1)*6 = 168。
# Conv2d-3 (880个)：(输入通道6 * 3*3 + 1) * 16 = (54+1)*16 = 880。
# Linear-5 (69,240个)：(输入特征 16*6*6=576) * 输出特征120 + 120个偏置 = 576*120 + 120 = 69,240。
# Linear-7 (850个)：84*10 + 10 = 850。
	train(train_dataset)
	#test(test_dataset)
````
