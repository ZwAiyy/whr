```python
"""
某城市一周最高气温为[28,30,29,31,32,30,29]。
-计算平均气温,最高气温和最低气温。
-找出气温超过30度的天数。
"""
```

```python
import numpy as np
temp = np.array([28,30,29,31,32,30,29])
avg_temp = np.mean(temp)
print(f'平均气温：{avg_temp:.3f}')
print(f'最高气温：{np.max(temp):.3f}')
print(f'最低气温：{np.min(temp):.3f}')

print(len(temp[temp>30]))
print(np.sum(temp>30))
print(np.count_nonzero(temp>30))
print(np.where(temp>30, temp , 0))
print(np.cumsum(np.where(temp>30, 1 , 0))[-1])
```

```python
'''A=[[1,2],[3,4]]和B=[[5,6],[7,8]]
计算A+B, A*B,矩阵乘法
'''
A=np.array([[1,2,2],[3,4,1],[3,4,2]])
B=np.array([[5,6,1],[7,8,4],[3,4,4]])
#print(A+B)
#print(A*B)
#print(A@B)
C=np.dot(A,B)
print(C)

C[C%2==0]=1
print(C)
```

```python
arr = np.arange(1,13)
arr = arr.reshape(3,4)
print(arr)

print('每行和',arr.sum(axis=1))
print('每列和',arr.sum(axis=0))
print('每行平均值',arr.mean(axis=1))
print('每列平均值',arr.mean(axis=0))
arr = arr.reshape(12)
arr = np.reshape(arr,12)
print(arr)
```

# Series

```python
import pandas as pd
s = pd.Series([1, 2, 3, 4, 5])
print(s)
#自定义索引
s = pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e'])
print(s)
#自定义name
s = pd.Series([1, 2, 3, 4, 5], index=['a', 'b', 'c', 'd', 'e'], name='num')
print(s)
```

```python
#通过字典创建
s = pd.Series({'a': 1, 'b': 2, 'c': 3})
print(s)

s1 = pd.Series(s, index=['b', 'c', 'd'])
print(s1)
```

```python
print(s.shape)
print(s.dtype)
print(s.size)
print(s.nbytes)
print(s.ndim)
print(s.flags, s.name, s.values)
print(s.loc['a'])#获取索引显示
print(s.iloc[0])#获取索引隐式
```

```python
import pandas as pd
import numpy as np
np.random.seed(42)
values = np.random.randint(50, 101 , 10)
index = []
for i in range(10):
    index.append('No.'+str(i+1))
score = pd.Series(values, index=index, name='score')
score.describe()
#高于平均分的人数
print('高于平均分的人数:',score[score>score.mean()].count())
```
